```python

# 1-The Task class is defined first because it represents the basic data unit that the system processes.
class Task:
    def __init__(self, task_id, description):
        self.task_id = task_id
        self.description = description # Stores the task description


# 2-Recommendation Class : represents an Agent recommendation
class Recommendation:
    def __init__(self, agent_name, content):
        self.agent_name = agent_name
        self.content = content

    def display(self)  :  # Method to display recommendation
        print(f"[{self.agent_name}] Recommendation:")
        print(self.content)
class Memory: # Main Logging System to ensure data continuity and performance monitoring
    def __init__(self):
        self.history = [  ]  # Empty registration list

    def log(self, task, agent_name, recommendation)  :# Method for data logging
        self.history.append(  {  # Adds a new item to the list
            "task": task.description,
            "agent": agent_name,
            "recommendation": recommendation
        }  )  # Close the dictionary

    def show_history(self)  :  # Another method to display all records
        print(" MEMORY HISTORY")
        print("- "* 10)
        for i, h in enumerate(self.history, 1)  :  # h represents a single transaction record, and i is the record numbering.
            print(f"{i}. Task: {h['task']}")
            print(f"Agent: {h['agent']}")
            print(f"Result: {h['recommendation']}")

# KnowledgeBase class
class KnowledgeBase:
    def __init__(self):
        self.data = {
            "users": 18000  ,  # Number of users
            "revenue": 7000  ,  # Revenue
            "expenses": 5000  ,  # Expenses
            "design_trends": ["Minimalist", "Luxury", "Modern"  ]  # Design trends
        }

    def get(self, key):#Method to get value
        return self.data.get(key)#The value is returned according to the key.

# Base Agent (Polymorphism :Same name, different task)
class Agent:
    def __init__(self, name, kb):
        self.name = name
        self.kb = kb

    def process_task(self, task):#A method to process the task
        return Recommendation(self.name, "Base response")


# Icon Identity Agent(first child)
class IconIdentityAgent(Agent):
    def __init__(self, name, kb):
        super().__init__(name, kb)#The constructor calls the father child

    def process_task(self, task):#Redefining (Polymorphism)
        trends = self.kb.get("design_trends")#Takes trends
        content = f"Suggested Design Styles: {', '.join(trends)}"#Converts the list to text
        return Recommendation(self.name, content)

# Growth Agent(The second child class)
class GrowthAgent(Agent):
    def __init__(self, name, kb):
        super().__init__(name, kb)

    def process_task(self, task):
        users = self.kb.get("users")#Brings the number of users

        if users >= 20000:
            content = "Start premium features now."
        else:
            content = f"Wait until 20,000 users (current: {users})."

        return Recommendation(self.name, content)

# Finance Agent(,The third child class ,Profit calculation)
class FinanceAgent(Agent):
    def __init__(self, name, kb):
        super().__init__(name, kb)

    def process_task(self, task):
        revenue = self.kb.get("revenue")
        expenses = self.kb.get("expenses")
        profit = revenue - expenses

        if profit > 0:
            content = f"Profit is positive: {profit}"
        else:
            content = f"Loss detected: {profit}"

        return Recommendation(self.name, content)

# Founder:This class represents the head of the system
class Founder:
    def __init__(self, name):
        self.name = name
        self.task_counter = 0#A counter that starts from zero to automatically number tasks.

    def create_task(self, description):#A function to receive and read a list of responses from customers.
        self.task_counter += 1
        task = Task(self.task_counter, description)
        print(f"Founder created task: {description}")
        return task

    def review(self, recommendations):#It goes through the list of recommendations and calls the display function for each one.
        print("REVIEWING RECOMMENDATIONS")
        for rec in recommendations:
            rec.display()

    def make_decision(self):#The final decision is based on the overall vision.
        print("FINAL DECISION")
        print("Focus on user growth before monetization.")
# MAIN SYSTEM :Program starting point.
def main():#The basic function that sums all the parts
    kb = KnowledgeBase()#Create a copy of the database
    memory = Memory()

    founder = Founder("Alaa Al-Shukailia")

    agents = [#Create a list containing the three client objects
        IconIdentityAgent("Icon Agent", kb),
        GrowthAgent("Growth Agent", kb),
        FinanceAgent("Finance Agent", kb)
    ]

    task = founder.create_task("When should we introduce premium features?")#The Founder sends the first mission.

    recommendations = []

    for agent in agents:#Rotation; it goes through each customer one after the other
        rec = agent.process_task(task)#It asks the current client to process the task and provide a response.
        recommendations.append(rec)
        memory.log(task, agent.name, rec.content)#The process is immediately recorded in memory to ensure it is not lost.

    founder.review(recommendations)
    founder.make_decision()
    memory.show_history()#Recalling memory at the end of the program to print the comprehensive report


if __name__ == "__main__":#The line that tells the computer: "Start the program from here"
    main()
```
